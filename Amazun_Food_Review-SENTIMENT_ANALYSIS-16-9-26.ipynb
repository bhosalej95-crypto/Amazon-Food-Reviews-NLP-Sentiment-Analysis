```python
import nltk
```

```python
from nltk.corpus import stopwords
```

```python
l1=stopwords.words('english')
l1
```

```python
import string
string.punctuation
```

```python
def text_process(mess):
    """
    1.remove the punctuatiom
    2.remove the stopwords
    3.return the list of clean textwords

    """
    nopunc=[char for char in mess if char not in string.punctuation]
    nopunc="".join(nopunc)
    return [word for word in nopunc.split() if word not in l1]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
# assign a Id/ token number to each word)
```

```python
bow_transformer =  CountVectorizer(analyzer = text_process).fit(ar['X'])
```

```python
len(bow_transformer.vocabulary_)
```

```python
# tdm -----> term document matrix
```

```python
tdm = bow_transformer.transform(ar['X'])
```

```python
tdm.shape
```

```python
type(tdm)
# tdm will be our X used in model building
```

```python
ar.Y=ar.Y.replace({1:1,2:1,3:2,4:2,5:2})
```

```python
from sklearn.model_selection import train_test_split
train_x, test_x, train_y,test_y = train_test_split(tdm,ar.Y)
```

```python
from sklearn.metrics import *
```

```python
from sklearn.linear_model import LogisticRegression
logrig=LogisticRegression()
logrig.fit(train_x,train_y)
pred = logrig.predict(test_x)
print(confusion_matrix(test_y,pred))
print(classification_report(test_y,pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier()
dt.fit(train_x, train_y)
pred = dt.predict(test_x)
tab = confusion_matrix(test_y, pred)
print(tab)
print(classification_report(test_y, pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, classification_report

rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    class_weight="balanced",
    n_jobs=-1,
    random_state=42
)

rf.fit(train_x, train_y)

pred = rf.predict(test_x)

print(confusion_matrix(test_y, pred))
print(classification_report(test_y, pred))
```
